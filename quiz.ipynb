### Create a bar chart showing various demographic breakdowns for Berkeley (race or gender)

```python
# Get the row for UC Berkeley
berkeley_data = df.loc[df['name'] == 'University of California-Berkeley'].iloc[0]

# Define the demographic columns related to race
race_demographics = [
    'american_indian_or_alaska_native',
    'asian',
    'black',
    'hispanic',
    'native_hawaiian_or_pacific_islander',
    'white',
    'two_or_more',
    'unknown'
]

# Extract the values for these columns
berkeley_race_counts = berkeley_data[race_demographics]

# Create the bar chart
plt.figure(figsize=(12, 7))
berkeley_race_counts.plot(kind='bar', color='skyblue')
plt.title('Racial Demographic Breakdown for UC Berkeley')
plt.xlabel('Race/Ethnicity')
plt.ylabel('Number of Students')
plt.xticks(rotation=45, ha='right') # Rotate labels for better readability
plt.tight_layout() # Adjust layout to prevent labels from overlapping
plt.show()
```

### Create several charts showing some interesting elements you find in these data

Some ideas might be comparing UC Berkeley to other UCs, looking at Ivy League comparisons, running some regressions

```python
ivy_league_schools = [
    'Brown University',
    'Columbia University in the City of New York',
    'Cornell University',
    'Dartmouth College',
    'Harvard University',
    'University of Pennsylvania',
    'Princeton University',
    'Yale University'
]

# Add UC Berkeley to the list for comparison
comparison_schools = ivy_league_schools + ['University of California-Berkeley']

# Filter the DataFrame for these schools
comparison_df = df[df['name'].isin(comparison_schools)].copy()

# Define the demographic columns related to race
race_demographics = [
    'american_indian_or_alaska_native',
    'asian',
    'black',
    'hispanic',
    'native_hawaiian_or_pacific_islander',
    'white',
    'two_or_more',
    'unknown'
]

# Calculate percentages for each demographic group
for demo_col in race_demographics:
    comparison_df[f'percent_{demo_col}'] = (comparison_df[demo_col] / comparison_df['enrollment']) * 100

# Select relevant columns for plotting (name and percentage columns)
plot_df = comparison_df[['name'] + [f'percent_{col}' for col in race_demographics]]
plot_df = plot_df.set_index('name')

# Rename columns for better readability on the chart
plot_df.columns = [
    'American Indian/Alaska Native',
    'Asian',
    'Black',
    'Hispanic',
    'Native Hawaiian/Pacific Islander',
    'White',
    'Two or More Races',
    'Unknown'
]

# Plotting
plt.figure(figsize=(15, 8))
plot_df.plot(kind='bar', figsize=(15, 8))
plt.title('Racial Demographic Breakdown: Ivy League vs. UC Berkeley')
plt.xlabel('University')
plt.ylabel('Percentage of Enrollment')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Race/Ethnicity', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
